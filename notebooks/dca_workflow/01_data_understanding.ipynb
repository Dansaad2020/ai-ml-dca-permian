## Data Understanding Notes

The dataset contains 1,220 normalized monthly production records for 30 Martin County wells. Production is normalized by month on production, which allows wells with different calendar start dates to be compared at the same life stage.

The full 30-well cohort is present through month 33. For the first model-comparison release, I will train models on months 1-24 and test on months 25-33. This avoids using future test-period data while keeping the test set within the full-cohort coverage window.

The first release will focus on oil production only. Gas forecasting is excluded for now because the prior project identified gas-fit boundary issues. This project is a model-comparison learning exercise, not a reserves report or representative Permian population study.
